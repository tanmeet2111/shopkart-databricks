# Create Catalog

```python
%sql
CREATE CATALOG IF NOT EXISTS shopkart;
```

# Create Three Schemas: Bronze, Silver, Gold

```python
%sql
CREATE SCHEMA IF NOT EXISTS shopkart.bronze;
CREATE SCHEMA IF NOT EXISTS shopkart.silver;
CREATE SCHEMA IF NOT EXISTS shopkart.gold;
```

# Create Volume in Bronze Schema

```python
%sql
CREATE VOLUME IF NOT EXISTS shopkart.bronze.raw_files;
```

```python
%sql
LIST "/Volumes/shopkart/bronze/raw_files";
```
